# NumPy Complete Practical Guide

This notebook covers NumPy from basics to advanced operations. Suitable for GitHub portfolio.

## 1. Importing NumPy

```
import numpy as np
print(np.__version__)
```

## 2. Creating Arrays

```

a = np.array([1, 2, 3])
b = np.array([[1, 2], [3, 4]])
c = np.zeros((2, 3))
d = np.ones((3, 3))
e = np.arange(0, 10, 2)
f = np.linspace(0, 1, 5)

a, b, c, d, e, f
```

## 3. Array Properties

```

print("Shape:", b.shape)
print("Dimensions:", b.ndim)
print("Size:", b.size)
print("Data type:", b.dtype)
```

## 4. Indexing and Slicing

```

arr = np.array([10, 20, 30, 40, 50])
print(arr[0])
print(arr[1:4])

mat = np.array([[1, 2, 3], [4, 5, 6]])
print(mat[0, 1])
print(mat[:, 1])
```

## 5. Array Operations

```

x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print(x + y)
print(x * y)
print(np.sqrt(y))
print(np.exp(x))
```

## 6. Broadcasting

```

mat = np.array([[1, 2, 3], [4, 5, 6]])
vec = np.array([10, 20, 30])
print(mat + vec)
```

## 7. Aggregations

```

data = np.array([[1, 2], [3, 4]])
print(np.sum(data))
print(np.mean(data))
print(np.max(data, axis=0))
```

## 8. Reshaping and Stacking

```

arr = np.arange(6)
print(arr.reshape(2, 3))

a = np.array([1, 2])
b = np.array([3, 4])
print(np.vstack((a, b)))
print(np.hstack((a, b)))
```

## 9. Boolean Masking

```

arr = np.array([1, 2, 3, 4, 5])
mask = arr > 3
print(arr[mask])
```

## 10. Linear Algebra

```

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print(np.dot(A, B))
print(np.linalg.det(A))
print(np.linalg.inv(A))
```

## 11. Random Module

```

np.random.seed(42)
print(np.random.rand(3))
print(np.random.randint(0, 10, size=5))
```

## 12. Saving and Loading Arrays

```

np.save("sample.npy", arr)
loaded = np.load("sample.npy")
print(loaded)
```

## 13. Practical Mini Example

```

# Normalize data
data = np.array([10, 20, 30, 40, 50])
normalized = (data - np.min(data)) / (np.max(data) - np.min(data))
print(normalized)
```
